## Counting streaks and probability simulations


```python
from collections import Counter

def count_streaks(xs):
    """Count number of success runs of length k."""
    s = Counter()

    curr_s = 0

    for x in xs:
        if x == 1:
            curr_s = curr_s + 1
        else:
            if curr_s > 0:
                s[curr_s] += 1
                curr_s = 0

    if curr_s > 0:
        s[curr_s] +=1


    return  s
```

```python
import numpy as np

def poisson_clt_simulator(sample_size, num_samples, lambda_):
    sample_means = []
    for _ in range(num_samples):
        sample = np.random.poisson(lam= lambda_, size=sample_size)
        sample_means.append(np.mean(sample)) # Think carefully what you are appending here, refer to variable name
    return sample_means
```

```python
import matplotlib.pyplot as plt

# Parameters
sample_size = 50
num_samples = 1000
lambda_ = 3

# Simulate and get sample means
sample_means = poisson_clt_simulator(sample_size,num_samples,lambda_)

# Plot the distribution of sample means
plt.hist(sample_means)

# Add labels and title
plt.xlabel('Sample Mean')
plt.ylabel('Density')
plt.title('Distribution of Sample Means (CLT Visualization for Poisson)')

# Show plot
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

size = 10000
n, p = 50, 0.5  # Large n for normal approximation
binomial_samples = np.random.binomial(n= n, p= p, size= size)
normal_samples = np.random.normal(loc=n*p, scale=np.sqrt(n*p*(1-p)), size=size) # Don't worry about this line unless you are interested

plt.hist(binomial_samples, bins=50, density=True, alpha=0.6, label="Binomial(50,0.5)")
plt.hist(normal_samples, bins=50, density=True, alpha=0.6, label="Normal Approximation")
plt.legend()
plt.title("Binomial vs. Normal Approximation")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```

## Campus setting and enrollment (ANOVA)


```python
# YOUR CODE HERE
urban = df[df["campusSetting"] == "Urban"]
suburban = df[df["campusSetting"] == "Suburban"]
rural = df[df["campusSetting"] == "Rural"]
```

```python
# YOUR CODE HERE
from scipy.stats import f_oneway

f_stat, p_value = f_oneway(urban["studentPopulation"], suburban["studentPopulation"], rural["studentPopulation"])
print(p_value)
```

```python
# YOUR CODE HERE
plt.boxplot([urban["studentPopulation"], suburban["studentPopulation"], rural["studentPopulation"]])

plt.xticks([1,2,3], ["Urban", "Suburban", "Rural"])
plt.xlabel("Campus Setting")
plt.ylabel("Student Population")
plt.title("Student Population by Campus Setting")
plt.show()
```

## Multiple comparisons and post-hoc analysis


```python
# Your code here
from scipy.stats import ttest_ind

alpha = 0.05 / 3

p1 = ttest_ind(urban["studentPopulation"], suburban["studentPopulation"]).pvalue
p2 = ttest_ind(urban["studentPopulation"], rural["studentPopulation"]).pvalue
p3 = ttest_ind(suburban["studentPopulation"],rural["studentPopulation"]).pvalue

print("Urban vs Suburban:", p1)
print("Urban vs Rural:", p2)
print("Suburban vs Rural:", p3)
print("Adjusted alpha:", alpha)
```
